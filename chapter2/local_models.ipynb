## Ollama

**make sure you've 1. installed ollama, downloaded (pulled) the model, and started "ollama serve" as explained in the book!**

```python
import sys
import os

sys.path.insert(0, os.path.abspath('..'))

from config import set_environment
from set_agents import get_agent

set_environment()
```

```python
# Using Ollama local model
chat = get_agent("ollama")
messages = [
    (
        "system",
        "You are a helpful assistant.",
    ),
    ("human", "What makes AI Engineering important?"),
]
ai_msg = chat.invoke(messages)
print(ai_msg.content)
```

## Huggingface

```python
from langchain_core.messages import SystemMessage, HumanMessage

chat_model = get_agent("huggingface")
messages = [
    SystemMessage(content="You're a helpful assistant"),
    HumanMessage(
        content="Explain the concept of machine learning in simple terms"
    ),
]
ai_msg = chat_model.invoke(messages)
print(ai_msg.content)
```
