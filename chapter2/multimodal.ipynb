## Image understanding

```python
from langchain_core.messages import HumanMessage
# from langchain_openai import ChatOpenAI

def analyze_image(image_url: str, question: str) -> str:
    chat = get_agent("openai")
    

    message = HumanMessage(
        content=[
            {
                "type": "text",
                "text": question
            },
            {
                "type": "image_url",
                "image_url": {
                    "url": image_url,
                    "detail": "auto"
                }
            }
        ]
    )

    response = chat.invoke([message])
    return response.content

# Example usage
image_url = "https://replicate.delivery/xezq/IKCEsW1cPI4YIJOzZSqI8hjSNu6A2Swu9sO8moagOK3zT3dF/tmpymxj5n13.webp"
questions = [
    "What objects do you see in this image?",
    "What is the overall mood or atmosphere?",
    "Are there any people in the image?"
]

for question in questions:
    print(f"\nQ: {question}")
    print(f"A: {analyze_image(image_url, question)}")
```
