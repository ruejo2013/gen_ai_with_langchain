# Prompt templates

```python
from langchain_core.prompts import FewShotPromptTemplate, PromptTemplate
from langchain_openai.chat_models import ChatOpenAI
from set_agents import get_agent
from langchain_core.output_parsers import StrOutputParser

# define once, reuse everywhere
question_template = PromptTemplate.from_template("Answer this question concisely: {question}")

question_with_context = PromptTemplate.from_template(
    """Context: {context}\n\nAnswer the following question based on the context above concisely: 
    {question}"""
)

chat = get_agent("openai")
question_msg = question_template.format(
    question="What is the capital of France?"
    )

context = input("Please say what context this AI Agent should assume:")
question = input("Please enter your question for the AI Agent:")  
chain = question_with_context| chat | StrOutputParser()
# Use the chain
result = chain.invoke({
    "context": f"{context}",
    "question": f"{question}"}
    )

print(result, type(result))
```
