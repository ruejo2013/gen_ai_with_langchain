For testing, there's the FakeListLLM

```python
from langchain_community.llms import FakeListLLM

# Create a fake LLM that always returns the same responses
fake_llm = FakeListLLM(responses=["Hello"])

result = fake_llm.invoke("Any input will return Hello")
print(result)  # Output: Hello
```

```python
from anthropic import Anthropic

client = Anthropic()

msg = client.messages.create(
    model="claude-3-5-haiku-20241022",
    max_tokens=50,
    messages=[{"role": "user", "content": "Hello"}],
)

print(msg.content[0].text)
```

The default interface to work with LLMs if the Chat interface though.

```python
import os
from langchain_anthropic import ChatAnthropic
from langchain_openai.chat_models import ChatOpenAI

from langchain_core.messages import SystemMessage, HumanMessage

### Use this for Anthropic chat model by
input_text = input("Please type the prompt here:")
# example text: "Write a Python function to calculate factorial"
# specifying a model:
# chat = ChatAnthropic(
#     model="claude-3-5-haiku-20241022",
#     api_key=os.environ.get("ANTHROPIC_API_KEY"),
#     temperature=0
# )
# # or:
chat = ChatOpenAI(model='gpt-5.1')

# messages:
messages = [
    SystemMessage(content="You're a helpful programming assistant"),
    HumanMessage(content=input_text)
]
response = chat.invoke(messages)
```

```python
print(response.content)
```

```python
def reverse_string(s: str) -> str:
    result = ''
    for ch in s:
        result = ch + result
    return result
```

```python
reverse_string("holidays break")
```
